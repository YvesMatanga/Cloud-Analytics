<h2>Telco Customer Churn Model Building</h2>

```python
import pandas as pd

X_train  = pd.read_csv('../data/input/X_train.csv')
y_train  = pd.read_csv('../data/input/y_train.csv')
X_test   = pd.read_csv('../data/input/X_test.csv')
y_test   = pd.read_csv('../data/input/y_test.csv')
```

<h2> Model Building</h2>

<h3>1. Logistic Regression</h3>

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

lg_clf = LogisticRegression()
lg_clf.fit(X_train, y_train)
y_pred_train_lg = lg_clf.predict(X_train)
y_pred_test_lg = lg_clf.predict(X_test)


print("[Logistic Regression]\nTrain - Classification report:\n", classification_report(y_train, y_pred_train_lg))
print("Test - Classification report:\n", classification_report(y_test, y_pred_test_lg))
```

<h3>2. Decision Trees</h3>

```python
from sklearn.tree import DecisionTreeClassifier

clf_dt = DecisionTreeClassifier(random_state=1234,
                                min_samples_split=2,max_depth=None,min_samples_leaf=1,criterion='gini')#model parameters
clf_dt.fit(X_train,y_train)

y_pred_train_dt = clf_dt.predict(X_train)
y_pred_test_dt = clf_dt.predict(X_test)

print("[Decision Trees]\nTrain - Classification report:\n", classification_report(y_train, y_pred_train_dt))
print("Test - Classification report:\n", classification_report(y_test, y_pred_test_dt))
```

<h3>3. Random Forest</h3>

```python
from sklearn.ensemble import RandomForestClassifier

clf_rf = RandomForestClassifier(random_state=1234,
                                n_estimators=100,min_samples_split=2,max_depth=None,min_samples_leaf=1,criterion='gini')#model parameters
clf_rf.fit(X_train,y_train)

y_pred_train_rf = clf_rf.predict(X_train)
y_pred_test_rf = clf_rf.predict(X_test)

print("[Random Forest]\nTrain - Classification report:\n", classification_report(y_train, y_pred_train_rf))
print("Test - Classification report:\n", classification_report(y_test, y_pred_test_rf))
```

<h3>4. Neural Networks: MLP </h3>

```python
from sklearn.model_selection import GridSearchCV
from sklearn.neural_network import MLPClassifier #Multi layer Perceptron


clf_mlp = MLPClassifier(max_iter=300,activation='relu',
                  early_stopping=False)

parameters_mlp = {'hidden_layer_sizes':[(30,), (30,20)]}

clf_mlp = GridSearchCV(clf_mlp,parameters_mlp,cv=5)
clf_mlp.fit(X_train,y_train)

y_pred_train_mlp = clf_rf.predict(X_train)
y_pred_test_mlp = clf_rf.predict(X_test)

print("[MLP]\nTrain - Classification report:\n", classification_report(y_train, y_pred_train_mlp))
print("Test - Classification report:\n", classification_report(y_test, y_pred_test_mlp))
```
